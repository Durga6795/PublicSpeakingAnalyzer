```python
#recent one with updated code for overal and all(this shld be made compact )
import os
import pickle
import cv2
import numpy as np
import librosa
import speech_recognition as sr
from pydub import AudioSegment, silence
from transformers import pipeline
from textblob import TextBlob
import mediapipe as mp
import time


class PublicSpeakingAnalyzer:
    def __init__(self, emotion_model_name="superb/hubert-large-superb-er"):
        print("🔧 Initializing fast analyzer...")
        self.mp_face = mp.solutions.face_mesh
        self.mp_hands = mp.solutions.hands
        self.mp_pose = mp.solutions.pose
        self.emotion_model_name = emotion_model_name
        self.emotion_model = pipeline("audio-classification", model=self.emotion_model_name)
        print("✅ Fast analyzer ready.")

    def analyze(self, video_path):
        start_time = time.time()
        if not os.path.exists(video_path):
            return {"Error": "❌ Video not found."}

        print(f"🎥 Processing video: {video_path}")

        # Extract audio
        os.system(f'ffmpeg -hide_banner -loglevel error -y -i "{video_path}" -vn -ac 1 -ar 16000 -f wav "temp.wav"')

        cap = cv2.VideoCapture(video_path)
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT)) or 1

        # Use only every 10th frame
        frame_interval = max(1, total_frames // 300)
        eye_frames = hand_frames = posture_frames = sampled = 0

        with self.mp_face.FaceMesh(max_num_faces=1, refine_landmarks=True, min_detection_confidence=0.5) as face, \
             self.mp_hands.Hands(max_num_hands=2, min_detection_confidence=0.5) as hands, \
             self.mp_pose.Pose(min_detection_confidence=0.5) as pose:

            while True:
                ret, frame = cap.read()
                if not ret:
                    break
                sampled += 1
                if sampled % frame_interval != 0:
                    continue
                rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

                # Eye contact detection
                if face.process(rgb).multi_face_landmarks:
                    eye_frames += 1

                # Gesture detection
                if hands.process(rgb).multi_hand_landmarks:
                    hand_frames += 1

                # Posture detection
                if pose.process(rgb).pose_landmarks:
                    posture_frames += 1

        cap.release()

        # Calculate video-based metrics
        eye_score = round(eye_frames / (sampled or 1) * 100, 2)
        gesture_score = round(hand_frames / (sampled or 1) * 100, 2)
        posture_score = round(posture_frames / (sampled or 1) * 100, 2)

        # ============================
        # 🎧 AUDIO ANALYSIS
        # ============================
        audio_file = "temp.wav"
        recognizer = sr.Recognizer()

        try:
            with sr.AudioFile(audio_file) as source:
                audio_data = recognizer.record(source)
            text = recognizer.recognize_google(audio_data)
        except:
            text = ""

        words = text.split()
        word_count = len(words)
        audio = AudioSegment.from_wav(audio_file)
        duration_minutes = len(audio) / (1000 * 60)
        pace_wpm = round(word_count / duration_minutes, 2) if duration_minutes else 0

        pauses = silence.detect_silence(audio, min_silence_len=400, silence_thresh=-40)
        total_pause = sum([p[1] - p[0] for p in pauses])
        pause_ratio = total_pause / len(audio)
        fluency_score = round((1 - pause_ratio) * 100, 2)

        # 🎙 Emotion detection (fast)
        emotion_result = self.emotion_model(audio_file)
        emotion = emotion_result[0]["label"]
        tone_conf = round(emotion_result[0]["score"] * 100, 2)

        # 📈 Pronunciation & confidence
        pronunciation_score = round(min(pace_wpm / 160 * 100, 100), 2)
        y, sr_ = librosa.load(audio_file, sr=None)
        energy = np.mean(librosa.feature.rms(y=y))
        confidence_score = round(min(energy * 400, 100), 2)

        mispronunciation = round(100 - pronunciation_score, 2)

        # 🧮 FIXED overall score
        metrics = [
            eye_score,
            gesture_score,
            posture_score,
            fluency_score,
            pronunciation_score,
            confidence_score
        ]
        overall_score = round(np.mean(metrics), 2)

        # 🗣 Suggestions
        suggestion = "Excellent delivery!"
        if overall_score < 70:
            suggestion = "Improve clarity and confidence."
        elif eye_score < 60:
            suggestion = "Try maintaining better eye contact."

        runtime = round(time.time() - start_time, 2)

        result = {
            "Eye Contact (%)": eye_score,
            "Hand Gestures (%)": gesture_score,
            "Posture (%)": posture_score,
            "Fluency (%)": fluency_score,
            "Pronunciation (%)": pronunciation_score,
            "Mispronunciation (%)": mispronunciation,
            "Pace (WPM)": pace_wpm,
            "Confidence (%)": confidence_score,
            "Emotion": emotion,
            "Tone Confidence (%)": tone_conf,
            "Overall Score (%)": overall_score,
            "Suggestion": suggestion,
            "Runtime (s)": runtime
        }

        print(f"✅ Analysis completed in {runtime} seconds.")
        return result

    # ============================
    # 💾 SAVE / LOAD SUPPORT
    # ============================
    def save(self, filename="public_speaking_model_fast.pkl"):
        config = {"emotion_model_name": self.emotion_model_name}
        with open(filename, "wb") as f:
            pickle.dump(config, f)
        print(f"✅ Model saved to {filename}")

    @classmethod
    def load(cls, filename="public_speaking_model_fast.pkl"):
        with open(filename, "rb") as f:
            config = pickle.load(f)
        if not isinstance(config, dict):
            raise TypeError("❌ Invalid model file")
        print("🔁 Reloading analyzer...")
        return cls(emotion_model_name=config["emotion_model_name"])


# ======================================================
# 🧩 USAGE EXAMPLE
# ======================================================
if __name__ == "__main__":
    analyzer = PublicSpeakingAnalyzer()
    analyzer.save("public_speaking_model_fast.pkl")

    # To test:
    # result = analyzer.analyze("sample.mp4")
    # print(result)
```

```python
# ============================================================
# 🎤 Multimodal Public Speaking Analyzer
# ============================================================
# This script analyzes a public speaking video using:
# - Video cues (eye contact, gestures, posture)
# - Audio cues (pace, fluency, tone via HuBERT ML model)
# - Emotion detection
# ============================================================

# ----------------------------
# 📦 Standard library imports
# ----------------------------

import os                       # file system operations
import time                     # timing execution
import math                     # mathematical operations
import uuid                     # generate unique temp filenames
from io import BytesIO          # handle audio bytes in memory

# ----------------------------
# 🧠 Third-party imports
# ----------------------------

import cv2                      # OpenCV for video processing
import numpy as np              # numerical computations
import librosa                  # audio resampling & processing
import soundfile as sf          # reading audio files

import torch                    # PyTorch base
import torch.nn as nn           # neural network layers

import mediapipe as mp          # MediaPipe for pose/face/hand analysis
import speech_recognition as sr # speech-to-text for WPM calculation

from transformers import (
    HubertModel,                # HuBERT backbone
    Wav2Vec2FeatureExtractor,   # feature extractor for HuBERT
    pipeline                    # HuggingFace pipeline (emotion)
)

# ============================================================
# 🔧 Device configuration
# ============================================================

# Automatically select GPU if available, otherwise CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ============================================================
# 🏷️ Fluency label configuration
# ============================================================

# Ordered fluency labels (best → worst)
LABELS = ["Excellent", "Good", "Moderate", "Poor"]

# Continuous fluency scoring weights for each label
FLUENCY_WEIGHTS = {
    "Poor": 0.25,
    "Moderate": 0.50,
    "Good": 0.75,
    "Excellent": 1.00
}

# ============================================================
# 🧠 HuBERT + Fluency Classifier Setup
# ============================================================

# Load HuBERT feature extractor (handles waveform → model input)
feature_extractor = Wav2Vec2FeatureExtractor.from_pretrained(
    "facebook/hubert-base-ls960"
)

# Load HuBERT model and move to device
hubert = HubertModel.from_pretrained(
    "facebook/hubert-base-ls960"
).to(device)

# Set HuBERT to inference mode
hubert.eval()

# ----------------------------
# 🎯 Fluency classifier network
# ----------------------------

class FluencyClassifier(nn.Module):
    """
    Feedforward neural network trained on HuBERT embeddings
    + meta features (pace, duration)
    """

    def __init__(self, input_dim=1538, num_classes=4):
        super().__init__()

        # Sequential fully-connected architecture
        self.net = nn.Sequential(
            nn.Linear(input_dim, 512),     # input → hidden
            nn.BatchNorm1d(512),           # stabilize training
            nn.ReLU(),                     # non-linearity
            nn.Dropout(0.25),              # regularization

            nn.Linear(512, 256),            # hidden → hidden
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.20),

            nn.Linear(256, num_classes)     # output layer
        )

    def forward(self, x):
        return self.net(x)

# Instantiate classifier and load trained weights
model = FluencyClassifier().to(device)
model.load_state_dict(
    torch.load("model/best_fluency_model.pt", map_location=device)
)

# Set classifier to inference mode
model.eval()

# ============================================================
# 🎧 ML-based Fluency Prediction Function
# ============================================================

def predict_fluency(audio_bytes, duration_sec):
    """
    Predicts:
    - Fluency label
    - Fluency score (0–1)
    - Tone score (0–1)
    - Pace (WPM)
    """

    # Read audio waveform from bytes
    audio, sample_rate = sf.read(BytesIO(audio_bytes), dtype="float32")

    # Resample audio to 16kHz if needed
    if sample_rate != 16000:
        audio = librosa.resample(
            audio,
            orig_sr=sample_rate,
            target_sr=16000
        )
        sample_rate = 16000

    # Normalize waveform amplitude
    audio = audio / (np.max(np.abs(audio)) + 1e-8)

    # ----------------------------
    # 🗣 Speech recognition for WPM
    # ----------------------------

    recognizer = sr.Recognizer()
    text = ""

    try:
        # NOTE: Google ASR is network-dependent and slow
        with sr.AudioFile(BytesIO(audio_bytes)) as src:
            audio_data = recognizer.record(src)
            text = recognizer.recognize_google(audio_data)
    except Exception:
        pass

    # Compute words per minute (pace)
    words = len(text.split())
    duration_min = max(1e-6, duration_sec / 60)
    pace = round(words / duration_min, 2)

    # ----------------------------
    # 🎼 HuBERT feature extraction
    # ----------------------------

    inputs = feature_extractor(
        audio,
        sampling_rate=16000,
        return_tensors="pt",
        padding=True
    )

    # Move inputs to GPU/CPU
    inputs = {k: v.to(device) for k, v in inputs.items()}

    # Forward pass through HuBERT
    with torch.no_grad():
        hidden = hubert(**inputs).last_hidden_state

    # Mean + Std pooling over time dimension
    emb = torch.cat(
        [hidden.mean(1), hidden.std(1)],
        dim=1
    )

    # Meta features: pace + duration
    meta = torch.tensor(
        [[pace, duration_min]],
        dtype=torch.float32
    ).to(device)

    # Concatenate HuBERT + meta features
    features = torch.cat([emb, meta], dim=1)

    # ----------------------------
    # 🎯 Fluency classification
    # ----------------------------

    with torch.no_grad():
        logits = model(features)
        probs = torch.softmax(logits, dim=1)[0]

    # Predicted class
    pred_idx = int(torch.argmax(probs))
    pred_label = LABELS[pred_idx]

    # Continuous fluency score (expected value)
    fluency_score = sum(
        probs[i].item() * FLUENCY_WEIGHTS[LABELS[i]]
        for i in range(len(LABELS))
    )

    # Tone score derived from entropy (confidence proxy)
    entropy = -sum(p.item() * math.log(p.item() + 1e-8) for p in probs)
    tone_score = 1 - entropy / math.log(len(LABELS))

    return {
        "label": pred_label,
        "pace_wpm": pace,
        "fluency_score": float(fluency_score),
        "tone_score": float(tone_score)
    }

# ============================================================
# 🎥 Public Speaking Analyzer (Main Class)
# ============================================================

class PublicSpeakingAnalyzer:
    """
    Combines:
    - Video cues (MediaPipe)
    - Audio ML fluency (HuBERT)
    - Emotion detection
    """

    def __init__(self, emotion_model_name="superb/hubert-large-superb-er"):
        print("🔧 Initializing PublicSpeakingAnalyzer...")

        # MediaPipe models
        self.mp_face = mp.solutions.face_mesh
        self.mp_hands = mp.solutions.hands
        self.mp_pose = mp.solutions.pose

        # Emotion recognition model (uses GPU if available)
        self.emotion_model = pipeline(
            "audio-classification",
            model=emotion_model_name,
            device=0 if torch.cuda.is_available() else -1
        )

        print("✅ Analyzer ready.")

    def analyze(self, video_path):
        start = time.time()

        # Validate video existence
        if not os.path.exists(video_path):
            return {"Error": "Video not found"}

        # Generate unique temp wav filename
        uid = uuid.uuid4().hex
        wav_path = f"temp_{uid}.wav"

        # ----------------------------
        # 🎧 Extract audio via FFmpeg
        # ----------------------------
        # NOTE: os.system used for simplicity
        os.system(
            f'ffmpeg -hide_banner -loglevel error -y '
            f'-i "{video_path}" -vn -ac 1 -ar 16000 "{wav_path}"'
        )

        # ----------------------------
        # 🎥 Video analysis (MediaPipe)
        # ----------------------------

        cap = cv2.VideoCapture(video_path)

        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT)) or 1
        fps = cap.get(cv2.CAP_PROP_FPS) or 30
        duration_sec = total_frames / fps

        # Sample ~300 frames max for speed
        frame_interval = max(1, total_frames // 300)

        eye = hands = posture = sampled = 0

        with self.mp_face.FaceMesh(refine_landmarks=True) as face, \
            self.mp_hands.Hands() as hand_model, \
            self.mp_pose.Pose() as pose:

            while True:
                ret, frame = cap.read()
                if not ret:
                    break

                sampled += 1
                if sampled % frame_interval != 0:
                    continue

                rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

                if face.process(rgb).multi_face_landmarks:
                    eye += 1
                if hand_model.process(rgb).multi_hand_landmarks:
                    hands += 1
                if pose.process(rgb).pose_landmarks:
                    posture += 1

        cap.release()

        # Normalize visual metrics
        eye_contact = round(eye / max(1, sampled), 2)
        hand_gestures = round(hands / max(1, sampled), 2)
        posture_score = round(posture / max(1, sampled), 2)

        # ----------------------------
        # 🎙 ML Fluency Analysis
        # ----------------------------

        with open(wav_path, "rb") as f:
            audio_bytes = f.read()

        fluency = predict_fluency(audio_bytes, duration_sec)

        # ----------------------------
        # 😊 Emotion analysis
        # ----------------------------

        emotion_out = self.emotion_model(wav_path)
        emotion = emotion_out[0]["label"]
        emotion_score = emotion_out[0]["score"]

        # ----------------------------
        # 🧮 Final scoring
        # ----------------------------

        confidence = round(
            (eye_contact + hand_gestures + posture_score) / 3, 2
        )

        overall = round(
            (
                confidence +
                fluency["fluency_score"] +
                fluency["tone_score"] * 0.8 +
                emotion_score * 0.8
            ) / 3.6,
            2
        )

        # Clean up temp file
        os.remove(wav_path)

        return {
            "Eye Contact": eye_contact,
            "Hand Gestures": hand_gestures,
            "Posture": posture_score,
            "Fluency Label": fluency["label"],
            "Fluency Score": round(fluency["fluency_score"], 2),
            "Tone": round(fluency["tone_score"], 2),
            "Pace (WPM)": fluency["pace_wpm"],
            "Emotion": emotion,
            "Confidence": confidence,
            "Overall Score": overall,
            "Runtime (s)": round(time.time() - start, 2)
        }

# ============================================================
# 🧪 TESTING BLOCK
# ============================================================

if __name__ == "__main__":
    analyzer = PublicSpeakingAnalyzer()

    video_path = "sample.mp4"  # <-- replace with your video path
    result = analyzer.analyze(video_path)

    print("\n📊 ANALYSIS RESULT")
    for k, v in result.items():
        print(f"{k}: {v}")


```
